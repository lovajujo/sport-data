```python
from statsbombpy import sb
comps=sb.competitions()
```

```python
target_id={
    "comp":53,
    "season":315
}
matches = sb.matches(competition_id=target_id["comp"], season_id=target_id["season"])
matches
```

```python
match_id=3998853
events=sb.events(match_id=match_id)
events.columns
```

```python
passes_dict={}
for match_id in matches["match_id"]:
    match_info = matches[matches["match_id"] == match_id].iloc[0]
    home_team = match_info["home_team"]
    away_team = match_info["away_team"]

    events=sb.events(match_id=match_id)
    passes_df = events[events["type"] == "Pass"].copy()
    passes_df = passes_df.dropna(subset=["location", "pass_end_location"])
    passes_df["start_x"] = passes_df["location"].apply(lambda x: x[0])
    passes_df["start_y"] = passes_df["location"].apply(lambda x: x[1])
    passes_df["end_x"] = passes_df["pass_end_location"].apply(lambda x: x[0])
    passes_df["end_y"] = passes_df["pass_end_location"].apply(lambda x: x[1])

    columns_to_keep = [
        "match_id", "team", "player", "pass_recipient",
        "start_x", "start_y", "end_x", "end_y", "pass_outcome"
    ]
    existing_columns = [col for col in columns_to_keep if col in passes_df.columns]
    passes_df = passes_df[existing_columns]

    passes_dict[match_id] = {}
    for team_name, team_df in passes_df.groupby("team"):
        passes_dict[match_id][team_name] = team_df

passes_dict
```

```python
# @title
passes = events[["team", "type", "location", "pass_end_location", "timestamp", "player"]]
passes=passes[passes["type"]=="Pass"]
passes
```

```python
# @title
import mplsoccer as mp
import matplotlib.pyplot as plt

def plot_passes(passes_df):
    start_x=passes_df["location"].apply(lambda x: x[0])
    start_y=passes_df["location"].apply(lambda x: x[1])
    end_x=passes_df["pass_end_location"].apply(lambda x: x[0])
    end_y=passes_df["pass_end_location"].apply(lambda x: x[1])
    pitch=mp.Pitch()
    fig, ax=pitch.draw()
    for i in start_x.index:
        pitch.arrows(start_x[i], start_y[i], end_x[i], end_y[i], width=0.5, color='0',  ax=ax)
    plt.show()
```

```python
# @title
nor_passes=passes[passes["team"]=="Norway Women's"]
plot_passes(nor_passes)
```

```python
import mplsoccer as mp
import matplotlib.pyplot as plt

def plot_passes(passes_df):
    pitch = mp.Pitch(pitch_type='statsbomb')
    fig, ax = pitch.draw()

    pitch.arrows(
        passes_df["start_x"], passes_df["start_y"],
        passes_df["end_x"], passes_df["end_y"],
        width=1, color='0', ax=ax
    )

    plt.show()

```
